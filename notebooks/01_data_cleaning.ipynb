## 2.3. Exploring Datasets

```python
# Verifying Datasets
print("Orders:", orders.shape)
print("Order Items:", order_items.shape)
print("Customers:", customers.shape)
print("Products:", products.shape)
print("Stores:", stores.shape)
print("County Population:", county_population.shape)
```

```python
# Basic Checks
def basic_checks(data, name):
    print("\n" + "=" * 35)
    print(f"DATASET: {name.upper()}")
    print("=" * 35)
    print("="*35, "SHAPE", "="*35)
    print(data.shape)
    print("="*35, "INFO", "="*35)
    data.info()
    print("="*35, "COLUMNS", "="*35)
    print(data.columns.to_list())
    print("="*35, "DATA TYPES", "="*35)
    print(data.dtypes)
    print("="*35, "NULLS", "="*35)
    print(data.isnull().sum())
    print("="*35, "MISSING VALUES PERCENTAGE", "="*35)
    print((data.isnull().sum()/len(data))*100)
    print("="*35, "DUPLICATES", "="*35)
    print(data.duplicated().sum())
    print("="*35, "UNIQUE VALUE COUNTS", "="*35)
    print(data.nunique())
    print("="*35, "NUMERICAL STATS", "="*35)
    print(data.describe().T)
   
```

```python
basic_checks(orders, "Orders")
basic_checks(order_items, "Order Items")
basic_checks(customers, "Customers")
basic_checks(products, "Products")
basic_checks(stores, "Stores")
basic_checks(county_population, "County Population")
```

## 2.4. Fix Data Types

### 2.4.1. Converting Date Columns

```python
# Convert date columns from strings to datetime

orders["order_date"] = pd.to_datetime(orders["order_date"])
orders["order_datetime"] = pd.to_datetime(orders["order_datetime"])

customers["registration_date"] = pd.to_datetime(customers["registration_date"])

stores["opening_date"] = pd.to_datetime(stores["opening_date"])
```

```python
# Verify conversions
print(orders[["order_date", "order_datetime"]].dtypes)
print(customers["registration_date"].dtype)
print(stores["opening_date"].dtype)
```

## 2.5. Handling Missing Values

### 2.5.1. Handling Missing Values in Orders

```python
# Count missing values before cleaning
payment_missing_before = orders["payment_method"].isnull().sum()

# Replace missing payment methods with Unknown
orders["payment_method"] = orders["payment_method"].fillna("Unknown")

# Check missing values after cleaning
payment_missing_after = orders["payment_method"].isnull().sum()

print("Before:", payment_missing_before)
print("After:", payment_missing_after)
```

### 2.5.2. Handling Missing Values in Customers

```python
# Missing age bands before cleaning
age_missing_before = customers["age_band"].isnull().sum()

# Fill missing age bands with Unknown
customers["age_band"] = customers["age_band"].fillna("Unknown")

# Missing age bands after cleaning
age_missing_after = customers["age_band"].isnull().sum()

print("Before:", age_missing_before)
print("After:", age_missing_after)
```


### 2.5.3 Handling Missing Unit Prices

```python
# Missing prices before cleaning
price_missing_before = order_items["unit_price_kes"].isnull().sum()

# Calculate missing unit prices
order_items["unit_price_kes"] = order_items["unit_price_kes"].fillna(
    (order_items["line_gross_kes"] / order_items["quantity"]).round(2)
)

# Missing prices after cleaning
price_missing_after = order_items["unit_price_kes"].isnull().sum()

print("Before:", price_missing_before)
print("After:", price_missing_after)
```

## 2.6 — Handling Duplicates

### 2.6.1 Removing Duplicate Records

```python
# Count rows and duplicates before cleaning
rows_before = len(order_items)
duplicates_before = order_items.duplicated().sum()

# Remove duplicate records
order_items = order_items.drop_duplicates()

# Check results after cleaning
rows_after = len(order_items)
duplicates_after = order_items.duplicated().sum()

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Duplicates before:", duplicates_before)
print("Duplicates after:", duplicates_after)
print("Rows removed:", rows_before - rows_after)
```

## 2.7. Handling Outliers

### 2.7.1 Identifying Quantity Outliers

```python
# Check how frequently each quantity occurs
order_items["quantity"].value_counts().sort_index()
```

```python
# Identify unusually large quantities
outliers = order_items[order_items["quantity"] >= 49].copy()

print("Potential outliers:", len(outliers))

# Inspect a few records
outliers[["line_id", "quantity", "unit_price_kes", "line_gross_kes"]].head()
```

```python
# Calculate quantities using gross sales and unit prices
outliers["calculated_quantity"] = (
    outliers["line_gross_kes"] / outliers["unit_price_kes"]
).round()

# Compare recorded and calculated quantities
outliers[["quantity", "calculated_quantity"]].head(10)
```

```python
# Check the calculated quantities for all outliers
print(outliers["calculated_quantity"].value_counts().sort_index())

# Count calculated quantities outside the expected range
print("Invalid calculated quantities:",
      (~outliers["calculated_quantity"].between(1, 5)).sum())
```

### 2.7.2 Correcting Quantity Outliers

```python
# Count outliers before correction
outliers_before = (order_items["quantity"] >= 49).sum()

# Replace incorrect quantities with calculated quantities
order_items.loc[outliers.index, "quantity"] = (
    outliers["calculated_quantity"].astype(int)
)

# Count outliers after correction
outliers_after = (order_items["quantity"] >= 49).sum()

print("Outliers before:", outliers_before)
print("Outliers after:", outliers_after)
print("Outliers corrected:", outliers_before - outliers_after)
```

```python
# Verify quantity distribution
order_items["quantity"].value_counts().sort_index()
```

## 2.8. Feature Engineering

### 2.8.1 Creating Date Features

```python
# Extract year and month from order_date
orders["year"] = orders["order_date"].dt.year
orders["month"] = orders["order_date"].dt.month

# Create year-month for monthly sales analysis
orders["year_month"] = orders["order_date"].dt.strftime("%Y-%m")

# Verify the new columns
orders[["order_date", "year", "month", "year_month"]].head()
```

# 2.9 Final Data Validation

### 2.9.1 Checking Quantity Consistency

```python
print("Total units in orders:", orders["item_units"].sum())

print("Total units in order items:", order_items["quantity"].sum())
```

### 2.9.2 Checking Quantity Consistency per Order

```python
# Calculate total quantities for each order
items_per_order = order_items.groupby("order_id")["quantity"].sum()

# Compare against the orders table
units_match = orders["item_units"] == orders["order_id"].map(items_per_order)

# Display results
print("Orders matching:", units_match.sum())
print("Orders not matching:", (~units_match).sum())
```

### 2.9.3 Checking Revenue Consistency

```python
# Calculate total revenue from both datasets
orders_revenue = orders["net_amount_kes"].sum()
items_revenue = order_items["line_net_kes"].sum()

# Compare the results
print("Revenue from orders:", round(orders_revenue, 2))
print("Revenue from order items:", round(items_revenue, 2))
print("Difference:", round(orders_revenue - items_revenue, 2))
```

### 2.9.4  Before and After Cleaning Summary

```python
print("Missing payment methods:", payment_missing_before, "->", payment_missing_after)
print("Missing age bands:", age_missing_before, "->", age_missing_after)
print("Missing unit prices:", price_missing_before, "->", price_missing_after)
print("Duplicate rows:", duplicates_before, "->", duplicates_after)
print("Quantity outliers:", outliers_before, "->", outliers_after)
```
